```python
import pandas as pd
import plotly.express as px

# Carregar os dados da BMW
tabela_bmw = pd.read_csv("bmw_global_sales_2018_2025.csv")

# Agrupar os dados por Ano e Região para somar a Receita
receita_ano_regiao = tabela_bmw.groupby(['Year', 'Region'])['Revenue_EUR'].sum().reset_index()

# Criar um gráfico de linhas interativo
grafico = px.line(receita_ano_regiao, 
                  x="Year", 
                  y="Revenue_EUR", 
                  color="Region",
                  title="Evolução da Receita por Região (2018-2025)",
                  labels={'Revenue_EUR': 'Receita em Euros', 'Year': 'Ano'})

# Mostrar o gráfico
grafico.show()
```

```python
import pandas as pd
import plotly.express as px

tabela_bmw = pd.read_csv("bmw_global_sales_2018_2025.csv")
# Carrega os dados da BMW.

transicao_eletrica = tabela_bmw.groupby('Year')['BEV_Share'].mean().reset_index()
# Calcula a média de elétricos por ano.

grafico_bev = px.bar(transicao_eletrica, x='Year', y='BEV_Share', 
                     title='Crescimento da Participação de Veículos Elétricos (BEV)',
                     labels={'BEV_Share': 'Participação Média (%)', 'Year': 'Ano'},
                     color='BEV_Share', 
                     color_continuous_scale='Blues',
                     text_auto='.2%') # Adiciona a porcentagem escrita em cima da barra
# Cria o gráfico de barras com as porcentagens visíveis.

# --- AJUSTE PARA NÃO CORTAR ---
grafico_bev.update_layout(
    margin=dict(l=50, r=50, t=80, b=100), # Aumenta o espaço nas bordas (Top e Bottom)
    xaxis=dict(type='category'),          # Garante que cada ano apareça como um nome, não um número corrido
    title_x=0.5                           # Centraliza o título para não cortar nas bordas
)

grafico_bev.show()
# Exibe o gráfico corrigido.
```

```python
# Vamos somar toda a receita gerada por cada região
receita_regional = tabela_bmw.groupby('Region')['Revenue_EUR'].sum().reset_index()
# Agrupa os dados por Região (Europa, China, USA, etc) e soma o valor total de vendas em Euros.

# Vamos calcular também o preço médio cobrado em cada região
preco_medio_regional = tabela_bmw.groupby('Region')['Avg_Price_EUR'].mean().reset_index()
# Calcula a média do preço dos carros vendidos em cada região para saber onde o público é mais "Premium".

grafico_receita = px.pie(receita_regional, values='Revenue_EUR', names='Region', 
                         title='Distribuição Global de Receita por Região',
                         hole=0.4)
# Cria um gráfico de "rosca" (pizza com furo) mostrando a fatia de mercado de cada continente.

grafico_receita.show()
# Exibe o gráfico. A maior fatia indica o mercado mais importante para o faturamento da empresa.
```

```python
import pandas as pd
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Carregar dados
tabela_bmw = pd.read_csv("bmw_global_sales_2018_2025.csv")

# Agrupar dados
df_tendencia = tabela_bmw.groupby('Year').agg({
    'Avg_Price_EUR': 'mean',
    'Units_Sold': 'sum'
}).reset_index()

# Criar o gráfico com subplots para evitar o erro de "Failed"
fig = make_subplots(specs=[[{"secondary_y": True}]])

# Adicionar Preço (Eixo 1)
fig.add_trace(
    go.Scatter(x=df_tendencia['Year'], y=df_tendencia['Avg_Price_EUR'], name="Preço Médio", mode='lines+markers'),
    secondary_y=False,
)

# Adicionar Vendas (Eixo 2)
fig.add_trace(
    go.Scatter(x=df_tendencia['Year'], y=df_tendencia['Units_Sold'], name="Unidades Vendidas", mode='lines+markers'),
    secondary_y=True,
)

# Ajustes de layout para não cortar
fig.update_layout(
    title_text="Análise de Tendência: Preço vs Vendas",
    margin=dict(l=50, r=50, t=80, b=50),
    xaxis=dict(type='category')
)

fig.show()
```

```python
# --- CÉLULA DE CONCLUSÃO E INSIGHTS ---

# 1. Calculando os números finais para o resumo
total_receita = tabela_bmw['Revenue_EUR'].sum()
# Soma todo o faturamento da BMW no período.

crescimento_bev = tabela_bmw[tabela_bmw['Year'] == 2025]['BEV_Share'].mean() - tabela_bmw[tabela_bmw['Year'] == 2018]['BEV_Share'].mean()
# Calcula quanto a participação de elétricos cresceu de 2018 até 2025.

regiao_lider = tabela_bmw.groupby('Region')['Revenue_EUR'].sum().idxmax()
# Identifica automaticamente qual região trouxe mais dinheiro.

# 2. Exibindo o Resumo Executivo
print(f"{'='*30}")
print("   RESUMO ESTRATÉGICO BMW   ")
print(f"{'='*30}")

print(f"1. FATURAMENTO TOTAL: A receita acumulada no período foi de EUR {total_receita:,.2f}.")
# Exibe o faturamento total formatado com vírgulas.

print(f"2. TRANSIÇÃO ELÉTRICA: Houve um crescimento de {crescimento_bev:.2f}% na participação de modelos BEV.")
# Mostra o avanço da eletrificação.

print(f"3. LIDERANÇA GLOBAL: A região de maior impacto financeiro foi: {regiao_lider}.")
# Informa o mercado principal.

print("\n--- RECOMENDAÇÃO DA ANÁLISE ---")
print("Com base nos dados, a BMW deve focar investimentos em infraestrutura de carregamento")
print(f"na região {regiao_lider}, onde a receita é alta e a aceitação de elétricos está em expansão.")
```
